```python
# =========================
# Customer Service: 10 prompts + GT Adherence
# - Shows results in Colab (local judge using your rubric)
# - Logs experiment + GT metric to Galileo UI
# =========================

!pip -q install --upgrade galileo openai pandas

import os, json, re
import pandas as pd
from openai import AzureOpenAI

from galileo import log
from galileo.datasets import create_dataset
from galileo.experiments import run_experiment
from galileo.schema.metrics import GalileoScorers

# -------------------------
# 0) HARD-CODED CONFIG (PASTE YOUR VALUES)

# os.environ["OPENAI_API_KEY"] = "6ffhCm6wvgxD2LRD7wNZI5sHgqHn4lqYOY7xn8Ycdg7vHvZ8qyujJQQJ99BCACYeBjFXJ3w3AAABACOGwuFA"
# os.environ["OPENAI_API_BASE"] = "https://tcoeaiteamgpt4o.openai.azure.com/"
# os.environ["OPENAI_API_VERSION"] = "2024-08-01-preview"
# os.environ["AZURE_DEPLOYMENT"] = "gpt-4o"

# # ============ Galileo ============
# os.environ["GALILEO_API_KEY"] = "SLXBE1sh1P1GRCoG4y5mPvbhGTdH2ScI1_tbBZbfQa0"
# # Project is optional in some flows, but strongly recommended
# os.environ["GALILEO_PROJECT"] = "galelio"
# -------------------------
GALILEO_API_KEY = "SLXBE1sh1P1GRCoG4y5mPvbhGTdH2ScI1_tbBZbfQa0"
GALILEO_PROJECT = "My first project"

AZURE_OPENAI_API_KEY = "6ffhCm6wvgxD2LRD7wNZI5sHgqHn4lqYOY7xn8Ycdg7vHvZ8qyujJQQJ99BCACYeBjFXJ3w3AAABACOGwuFA"
AZURE_OPENAI_ENDPOINT = "https://tcoeaiteamgpt4o.openai.azure.com/"
AZURE_OPENAI_API_VERSION ="2024-08-01-preview"
AZURE_DEPLOYMENT ="gpt-4o"

# If you duplicated/created a custom metric in Galileo UI for (format + meaning),
# put its EXACT name here and set USE_CUSTOM_GALILEO_METRIC=True.
USE_CUSTOM_GALILEO_METRIC = False
CUSTOM_GALILEO_METRIC_NAME = "Ground Truth Adherence - Copy"  # example

# Galileo SDK usually reads env vars; set them for reliability
os.environ["GALILEO_API_KEY"] = GALILEO_API_KEY
os.environ["GALILEO_PROJECT"] = GALILEO_PROJECT

# -------------------------
# 1) Customer-service prompt template (forces JSON format)
# -------------------------
SYSTEM_PROMPT = """You are a customer support agent.
Respond ONLY as valid JSON with exactly these keys:
{
  "classification": "BILLING|SHIPPING|REFUND|ACCOUNT|TECH",
  "apology": "...",
  "resolution": "...",
  "next_steps": "...",
  "policy_ref": "..."
}
Rules:
- Do not add extra keys.
- Keep each field concise.
"""

# -------------------------
# 2) Create 10 prompts + ground truth (domain-specific)
# Ground truth MUST be placed in dataset 'output' for GT adherence.
# -------------------------
data = [
  {
    "input": "Customer: I was charged twice for order #A123. Please fix this.",
    "ground_truth": {"classification":"BILLING","apology":"Sorry about the duplicate charge.","resolution":"I can help reverse the extra charge for order #A123.","next_steps":"Please share the last 4 digits of the card and the charge amounts/dates so I can locate both transactions and initiate a reversal.","policy_ref":"Billing corrections: duplicate charges are refunded to the original payment method."}
  },
  {
    "input": "Customer: My package says delivered but I don't have it. Order #B552.",
    "ground_truth": {"classification":"SHIPPING","apology":"Sorry you haven’t received your order.","resolution":"I’ll help investigate the delivery for order #B552.","next_steps":"Please confirm your delivery address and check with neighbors/building staff; I’ll start a carrier trace and arrange a replacement or refund if confirmed lost.","policy_ref":"Delivered-not-received: we open a carrier investigation and offer replacement/refund if unresolved."}
  },
  {
    "input": "Customer: I want a refund. The shoes don't fit and I received them yesterday.",
    "ground_truth": {"classification":"REFUND","apology":"Sorry the fit wasn’t right.","resolution":"You’re eligible to return the shoes for a refund since they were received yesterday.","next_steps":"Please confirm your order number; I’ll send a return label and process the refund to the original payment method after the return is received.","policy_ref":"Returns: items can be returned within the return window for a refund once received back."}
  },
  {
    "input": "Customer: I can't log in—my password reset email never arrives.",
    "ground_truth": {"classification":"ACCOUNT","apology":"Sorry you’re having trouble signing in.","resolution":"Let’s get your password reset working.","next_steps":"Check spam/junk and confirm the email address; if it still doesn’t arrive, I can verify your identity and help update the email or trigger a manual reset.","policy_ref":"Account access: identity verification is required for email changes; reset emails may be delayed."}
  },
  {
    "input": "Customer: The app crashes when I open it on Android 14.",
    "ground_truth": {"classification":"TECH","apology":"Sorry about the crashing issue.","resolution":"We’ll troubleshoot the Android 14 crash.","next_steps":"Share your device model and app version; try force close, clear cache, and reinstall. If it persists, I’ll escalate to engineering with your details.","policy_ref":"Technical issues: collect device/app details and escalate reproducible crashes."}
  },
  {
    "input": "Customer: Can you change my shipping address? I placed the order 10 minutes ago.",
    "ground_truth": {"classification":"SHIPPING","apology":"Sorry for the inconvenience.","resolution":"I can try to update the shipping address since the order was placed recently.","next_steps":"Send the order number and full new address; if fulfillment already started we may need to intercept or reship.","policy_ref":"Address changes: possible only before fulfillment begins; otherwise intercept/reship options apply."}
  },
  {
    "input": "Customer: I was promised free shipping but I got charged for it.",
    "ground_truth": {"classification":"BILLING","apology":"Sorry you were charged shipping unexpectedly.","resolution":"I’ll verify the offer and correct the shipping charge if you were eligible.","next_steps":"Share your order number and promo details (code/screenshot); I’ll confirm eligibility and refund the shipping fee if applicable.","policy_ref":"Promotions: shipping fees are refunded when the order qualifies under the free-shipping terms."}
  },
  {
    "input": "Customer: Please cancel my subscription immediately.",
    "ground_truth": {"classification":"ACCOUNT","apology":"Sorry to see you go.","resolution":"I can help cancel your subscription renewal.","next_steps":"Confirm the email on the subscription; I’ll cancel renewal and tell you the end date of access based on your billing cycle.","policy_ref":"Subscriptions: cancellation stops renewal; access continues until the end of the paid period."}
  },
  {
    "input": "Customer: My refund is taking too long. It's been 12 days.",
    "ground_truth": {"classification":"REFUND","apology":"Sorry the refund is taking longer than expected.","resolution":"I’ll check the refund status and timeline.","next_steps":"Share your order number and refund confirmation date; I’ll verify whether it’s pending, processed, or delayed by your bank and give the next expected timeline.","policy_ref":"Refunds: processing times vary by payment method and bank; we can track status once initiated."}
  },
  {
    "input": "Customer: The discount code says invalid but it should work.",
    "ground_truth": {"classification":"BILLING","apology":"Sorry the discount code isn’t applying.","resolution":"Let’s identify why the code shows as invalid.","next_steps":"Share the exact code and items in your cart; I’ll check eligibility rules (expiration, exclusions, minimum spend) and suggest a fix or alternative.","policy_ref":"Discounts: codes may be limited by dates, exclusions, and cart requirements."}
  },
]

# Store ground truth as JSON text (stable comparisons)
df = pd.DataFrame([{"input": d["input"], "ground_truth": json.dumps(d["ground_truth"], ensure_ascii=False)} for d in data])

# -------------------------
# 3) Azure OpenAI client
# -------------------------
client = AzureOpenAI(
    api_key=AZURE_OPENAI_API_KEY,
    api_version=AZURE_OPENAI_API_VERSION,
    azure_endpoint=AZURE_OPENAI_ENDPOINT,
)

# -------------------------
# 4) Model under test (always returns a STRING; never None)
# This fixes the UI Output being "—".
# -------------------------
@log
def model_runner(row: dict) -> str:
    prompt = row["input"]
    try:
        resp = client.chat.completions.create(
            model=AZURE_DEPLOYMENT,
            messages=[
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": prompt},
            ],
            temperature=0.0,
        )
        out = resp.choices[0].message.content
        if not out or not out.strip():
            return "[ERROR] Empty model response"
        return out.strip()
    except Exception as e:
        return f"[ERROR] Exception during model call: {str(e)}"

# -------------------------
# 5) LOCAL JUDGE (your rubric: format + meaning)
# Shows results in Colab.
# -------------------------
JUDGE_SYSTEM_PROMPT = """You are evaluating whether the output is equivalent to the provided ground truth text.

Carefully compare the output and the ground truth, identifying both similarities and differences. For each difference you find, analyze whether it affects the meaning of the text. Clearly explain your reasoning for each difference, considering if it changes the conclusions a reasonable reader would draw. After examining all differences, decide whether the meaning is fully preserved or not.

Format requirement (MUST enforce):
- The output MUST be valid JSON.
- The output MUST contain exactly these keys: classification, apology, resolution, next_steps, policy_ref.
- No extra keys. classification must be one of BILLING|SHIPPING|REFUND|ACCOUNT|TECH.
If format fails, mark as False.

Meaning requirement:
- Accept differences in phrasing/wording as long as meaning is fully preserved.
- Any change in key facts, scope, or intent that could change conclusions => False.

Rubric:
- True: Equivalent meaning + format valid.
- False: Any meaning change OR format invalid.
"""

JUDGE_OUTPUT_FORMAT = """Return ONLY valid JSON:
{"equivalent": true/false, "reasoning": "brief reason"}"""

def _extract_json_obj(text: str):
    text = text.strip()
    try:
        return json.loads(text)
    except Exception:
        m = re.search(r"\{.*\}", text, flags=re.DOTALL)
        if not m:
            raise ValueError(f"Not JSON: {text[:300]}")
        return json.loads(m.group(0))

def local_judge(ground_truth_json_text: str, model_output_text: str):
    user = f"""GROUND TRUTH (JSON TEXT):
{ground_truth_json_text}

MODEL OUTPUT:
{model_output_text}

{JUDGE_OUTPUT_FORMAT}
"""
    resp = client.chat.completions.create(
        model=AZURE_DEPLOYMENT,
        messages=[
            {"role": "system", "content": JUDGE_SYSTEM_PROMPT},
            {"role": "user", "content": user},
        ],
        temperature=0.0,
    )
    raw = resp.choices[0].message.content.strip()
    obj = _extract_json_obj(raw)
    eq = obj.get("equivalent", False)
    if isinstance(eq, str):
        eq = eq.strip().lower() in ("true", "yes", "1")
    return bool(eq), obj.get("reasoning", raw)

# -------------------------
# 6) RUN LOCAL EVAL (Colab results table)
# -------------------------
local_rows = []
for r in df.to_dict(orient="records"):
    out = model_runner({"input": r["input"]})  # direct call, same logic as Galileo run
    eq, reason = local_judge(r["ground_truth"], out)
    local_rows.append({
        "input": r["input"],
        "ground_truth": r["ground_truth"],
        "model_output": out,
        "gt_adherence_pass": eq,
        "judge_reason": reason
    })

local_res = pd.DataFrame(local_rows)
score = local_res["gt_adherence_pass"].mean() * 100
print(f"\n✅ LOCAL Ground Truth Adherence (Format+Meaning) Pass Rate: {score:.2f}%\n")
local_res

# -------------------------
# 7) CREATE GALILEO DATASET (input/output/metadata)
# Ground truth must be in output.
# -------------------------
dataset_name = "cs-gt-adherence-10"
content = []
for r in df.to_dict(orient="records"):
    content.append({
        "input": r["input"],
        "output": r["ground_truth"],   # ground truth here
        "metadata": {"domain": "customer_support"},
    })

dataset = create_dataset(name=dataset_name, content=content)
print("\n✅ Galileo dataset created:", dataset_name, "| id:", dataset.id)

# -------------------------
# 8) Pick ONLY the Galileo metric
# - If you have a custom metric in UI, use it by name string.
# - Else auto-detect built-in Ground Truth Adherence if present.
# -------------------------
if USE_CUSTOM_GALILEO_METRIC:
    galileo_metrics = [CUSTOM_GALILEO_METRIC_NAME]
    print("✅ Using CUSTOM Galileo metric:", CUSTOM_GALILEO_METRIC_NAME)
else:
    # Try to find a built-in scorer for ground truth adherence in this SDK
    attrs = [a for a in dir(GalileoScorers) if not a.startswith("_")]
    candidates = ["ground_truth_adherence", "groundtruth_adherence"]
    metric_obj = None
    used_name = None
    for c in candidates:
        if hasattr(GalileoScorers, c):
            metric_obj = getattr(GalileoScorers, c)
            used_name = c
            break
    if metric_obj is None:
        # Fuzzy match fallback
        fuzzy = [a for a in attrs if ("ground" in a.lower() and "truth" in a.lower() and ("adher" in a.lower() or "align" in a.lower()))]
        if fuzzy:
            used_name = fuzzy[0]
            metric_obj = getattr(GalileoScorers, used_name)

    if metric_obj is None:
        print("\n⚠️ Could not auto-find built-in Ground Truth Adherence scorer in your GalileoScorers.")
        print("Available scorers:", attrs)
        print("➡️ Set USE_CUSTOM_GALILEO_METRIC=True and use your custom metric name from Galileo UI.")
        raise RuntimeError("No Ground Truth Adherence scorer found in this SDK.")
    galileo_metrics = [metric_obj]
    print("✅ Using built-in Galileo scorer:", used_name)

# -------------------------
# 9) Run Galileo experiment (logs to UI)
# -------------------------
experiment_name = "cs-gt-adherence-only"

results = run_experiment(
    experiment_name=experiment_name,
    dataset=dataset,
    function=model_runner,
    metrics=galileo_metrics,  # ONLY ONE metric
    project=GALILEO_PROJECT,
)

print("\n✅ GALILEO EXPERIMENT COMPLETED:", experiment_name)
print(results)
print("\n➡️ Galileo UI → Project → Experiments →", experiment_name)
```
